#  Skill Feature Strategy Comparison for Classification


## ⚙️ Strategy 1: Full One-Hot Encoding + PCA

1. Apply One-Hot Encoding on all 34 unique skills  
2. Concatenate encoded features with other data  
3. Standardize features  
4. Apply PCA  
5. Train classification model  
6. Evaluate performance

---

## ⚙️ Strategy 2: Grouped Skills + (Optional) PCA

1. Manually group skills into ~11–12 categories  
2. One-hot encode grouped categories  
3. Concatenate with other features  
4. Optionally apply PCA  
5. Train classification model  
6. Evaluate performance

```python
# skill groups by keywords
tools = ['Power Bi Reports', 'Oracle Fusion Finance', 'Microstrategy', 'Dynamics 365', 'ERP', 'R12']
analytical = ['analytical', 'data analyses', 'equity research', 'business analysis', 'analysis']
reporting = ['financial analysis', 'financial forecasting', 'financial reporting', 
             'financial data analysis', 'financial statements']
planning = ['financial planning', 'financial analyst', 'associate financial analyst', 
            'structured finance', 'investment banking', 'finance']
accounting = ['general ledger', 'due diligence', 'assurance', 'kyc']
management = ['analyst', 'team management', 'post production', 'automation']
```

```python
# Function to assign each skill to a group
def group_skill(skill):
    skill_lower = skill.lower()
    if skill in tools:
        return skill  # keep tool name as group
    elif any(k in skill_lower for k in analytical):
        return 'Analytical & Research Skills'
    elif any(k in skill_lower for k in reporting):
        return 'Financial Analysis & Reporting'
    elif any(k in skill_lower for k in planning):
        return 'Financial Planning & Strategy'
    elif any(k in skill_lower for k in accounting):
        return 'Accounting & Compliance'
    elif any(k in skill_lower for k in management):
        return 'Management & Role-Based'
    else:
        return 'Other'
```

```python
# Apply the mapping to our dataframe
df['Skills'] = df['Skills'].apply(group_skill)
```

```python
# Function to categorize job titles
def standardize_title(title):
    title = title.lower()
    roles = []

    if 'manager' in title :
        roles.append('Manager')
    if 'manager' in title:
        roles.append('Manager')
    if 'senior' in title or 'lead' in title:
        roles.append('Senior Analyst')
    if 'business analyst' in title:
        roles.append('Business Analyst')
    if 'consultant' in title:
        roles.append('Consultant')
    if 'analyst' in title and 'Senior Analyst' not in roles and 'Business Analyst' not in roles:
        roles.append('Analyst')

    # Ensure unique, readable and sorted titles
    return "/".join(sorted(set(roles))) if roles else "Unknown"
```

```python
# Apply the mapping to our dataframe
df['Title'] = df['Title'].apply(standardize_title)
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold,cross_validate
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import RandomOverSampler,SMOTE
from sklearn.model_selection import GridSearchCV
```

```python
# skills_to_remove = ['Microstrategy', 'R12', 'Dynamics 365', 'Power Bi Reports']
skills_to_remove = [2,3,7,9,10]

# Filter out the rows from X and y where the target skill is in the 'skills_to_remove' list
X_filtered = X[~np.isin(y, skills_to_remove)]  # Filter rows in X where y (Skills) is not in skills_to_remove
y_filtered = y[~np.isin(y, skills_to_remove)]  # Similarly filter y
```

```python
pd.DataFrame(X).shape
```

```python
# A copy for xgboost and lgboost models

le2 = LabelEncoder()

y_filtered_copy = y_resampled.copy()
y_filtered_encoded = le2.fit_transform(y_filtered_copy)
pd.DataFrame(y_filtered_encoded).value_counts()
```
